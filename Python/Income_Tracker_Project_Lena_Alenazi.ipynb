```python
import json
import os
```

```python
# Define a transaction class
class Transaction:
  def __init__(self, amount, category, date, type="expense"):
    self.amount = amount
    self.category = category
    self.date = date
    self.type = type  # "income" or "expense"
```

```python
# Function to add a transaction
def add_transaction(transactions, amount, category, date, type="expense"):
  transaction = Transaction(amount, category, date, type)
  transactions.append(transaction)
  return transactions
```

```python
# Function to view transactions
def view_transactions(transactions, sort_by="date"):
  if sort_by == "date":
    transactions.sort(key=lambda t: t.date)
  elif sort_by == "category":
    transactions.sort(key=lambda t: t.category)
  for transaction in transactions:
    print(f"{transaction.type.capitalize()}: ${transaction.amount} - {transaction.category} ({transaction.date})")
```

```python
# Function to update transactions
def update_transaction(transactions, index):
    transaction = transactions[index - 1]  # Adjust index to 0-based
    amount = float(input("Enter new amount: "))
    category = input("Enter new category: ")
    date = input("Enter new date (YYYY-MM-DD): ")
    transaction.amount = amount
    transaction.category = category
    transaction.date = date
```

```python
# Function to delete a transaction
def delete_transaction(transactions, index):
    del transactions[index - 1]  # Adjust index to 0-based
```

```python
# Function to generate basic reports
def generate_reports(transactions):
  total_income = sum(t.amount for t in transactions if t.type == "income")
  total_expense = sum(t.amount for t in transactions if t.type == "expense")
  category_breakdown = {}
  for transaction in transactions:
    category_breakdown[transaction.category] = category_breakdown.get(transaction.category, 0) + transaction.amount
  avg_transaction_value = {}
  for category, total in category_breakdown.items():
    count = sum(1 for t in transactions if t.category == category)
    avg_transaction_value[category] = total / count if count > 0 else 0

  print(f"Total Income: ${total_income}")
  print(f"Total Expense: ${total_expense}")
  print("Spending by Category:")
  for category, amount in category_breakdown.items():
    print(f"{category}: ${amount}")
  print("Average Transaction Value by Category:")
  for category, avg in avg_transaction_value.items():
    print(f"{category}: ${avg:.2f}")
```

```python
# Function to save transactions to a file
def save_transactions(transactions, username):
  filename = f"transactions_{username}.json"
  with open(filename, "w") as f:
    json.dump([vars(t) for t in transactions], f, indent=2)

# Function to load transactions from a file
def load_transactions(username):
  filename = f"transactions_{username}.json"
  transactions = []
  if not os.path.exists(filename):
    print(f"File '{filename}' does not exist.")
  else:
    try:
      with open(filename, "r") as f:
        data = json.load(f)
        transactions = [Transaction(**t) for t in data]
    except FileNotFoundError:
      pass
  return transactions
```

```python

# Main program loop
username = input("Enter your username: ")
transactions = load_transactions(username)

while True:
    print("\nMenu:")
    print("1. Add Transaction")
    print("2. View Transactions")
    print("3. Generate Reports")
    print("4. Save Transactions")
    print("5. Load Transactions")
    print("6. Update Transaction")
    print("7. Delete Transaction")
    print("8. Exit")
    choice = input("Enter your choice: ")

    if choice == "1":
        amount = float(input("Enter amount: "))
        category = input("Enter category: ")
        date = input("Enter date (YYYY-MM-DD): ")
        type = input("Enter transaction type (income/expense): ").lower()
        transactions = add_transaction(transactions, amount, category, date, type)
    elif choice == "2":
        sort_by = input("Sort by (date/category): ").lower()
        view_transactions(transactions, sort_by)
    elif choice == "3":
        generate_reports(transactions)
    elif choice == "4":
        save_transactions(transactions, username)
    elif choice == "5":
        transactions = load_transactions(username)
        if not transactions:
            print("No transactions found.")
    elif choice == "6":
        index = int(input("Enter index of transaction to update: "))
        update_transaction(transactions, index)
    elif choice == "7":
        index = int(input("Enter index of transaction to delete: "))
        delete_transaction(transactions, index)
    elif choice == "8":
        break
    else:
        print("Invalid choice!")

print("Exiting program...")
```
